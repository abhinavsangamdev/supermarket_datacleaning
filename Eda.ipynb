```python
#Which Category generates the highest total Sales?
import pandas as pd 
import numpy as np 
df = pd.read_csv("superstore_cleaned.csv")
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

```python
#Which Sub-Category generates the highest Sales?
df.groupby("Sub-Category")['Sales'].sum().sort_values(ascending=False)
```

```python
# Sales by region
df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
```

```python
# Which Category generates the highest Profit?
df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
```

```python
#Region-wise Profit
df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
```

```python
df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
```

```python
#Discount vs Profit
df.groupby('Discount')['Profit'].sum().sort_values(ascending=False)
```

```python
#solving that negative values 
df.groupby('Discount')['Profit'].agg(['mean','sum'])
```

```python
#Region + Category
df.groupby(['Region','Category'])['Profit'].sum().sort_values(ascending=False)
```
