```python

df['Ship Mode'] = df['Ship Mode'].fillna(df['Ship Mode'].mode()[0])
#cleaned the data of Ship Mode that is in string
```

```python

df['Customer Name'] = df['Customer Name'].fillna("unknown")
# to not keep the culumn empty we have named all as unknown
```

```python

df['Sales'] = df['Sales'].fillna(df['Sales'].median())
# we filled sales by the median values 
```

```python

df = df.dropna(subset = ["Postal Code"] )
print(df["Postal Code"].isnull().sum())
#deleted the postal code because we cannot assign random number to that
```

```python
df.drop_duplicates(inplace=True)
# we have successfully droped duplicates 
```

```python
for col in ['Ship Mode', 'Segment', 'Category', 'Sub-Category', 'Region']:
    print(col)
    print(df[col].unique())
    print()
```
